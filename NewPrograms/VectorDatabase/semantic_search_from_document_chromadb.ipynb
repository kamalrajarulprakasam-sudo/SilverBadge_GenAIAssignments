```python
import chromadb

document = """Employees can take up to 20 days of paid vacation each year.
Submit vacation requests through the HR portal at least two weeks in advance.

Employees can work from home up to two days per week.
They should agree on their remote work schedule with their manager.

The company reimburses up to ₹15,000 per year for courses and certifications.
Employees must get manager approval before enrolling."""

# Split the document into paragraphs, one per topic.
passages = document.split("\n\n")

# Create an in-memory ChromaDB collection.
client = chromadb.EphemeralClient()
collection = client.get_or_create_collection(name="employee_handbook")

# ChromaDB embeds each passage and stores its vector alongside the text.
collection.upsert(
    ids=["vacation", "remote_work", "training"],
    documents=passages,
)

question = input("Ask a question about the employee handbook: ").strip()

if question:
    # ChromaDB embeds the question and finds the closest passage.
    results = collection.query(
        query_texts=[question],
        n_results=1,
    )

    best_passage = results["documents"][0][0]
    print("\nAnswer from the document:")
    print(best_passage)
else:
    print("Please enter a question.")
```
